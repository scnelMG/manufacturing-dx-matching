# 공개용 노트북 안내

이 폴더는 2025년 산학협력 프로젝트의 실제 처리 흐름을 개인정보·원본 데이터·실행 출력 없이 정리한 포트폴리오용 코드입니다. 모든 경로는 비공개 입력과 생성 산출물을 가리키며, 이 저장소만으로 실제 기업 매칭 결과를 재생성할 수는 없습니다.

## 실행 순서

1. `01_task_check.ipynb` - 비공개 기업 마스터 입력 계약 검증
2. `02_company_preprocessing.ipynb` - 수요기업 검색 문장 생성
3. `03_homepage_description_generation.ipynb` - 선택 사항: 홈페이지 본문 기반 설명 보강
4. `04_technology_data_preprocessing.ipynb` - 특허·제품·솔루션 기술 문서 통합
5. `05_technology_embedding.ipynb` - KR-SBERT 기술 임베딩 생성
6. `06_company_matching.ipynb` - 코사인 유사도 검색과 공급기업 재정렬

`03`의 현재 SDK 구현은 2025년의 역사적 LLM 출력과 동일한 문장을 재현한다는 보장이 없습니다. 결과 해석에는 `results/`의 실제 당시 산출물을 사용합니다.
