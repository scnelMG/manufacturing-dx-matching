# 공개용 노트북 안내

이 폴더는 2025년 산학협력 프로젝트의 실제 처리 흐름을 개인정보·원본 데이터·실행 출력 없이 정리한 포트폴리오용 코드입니다. 모든 경로는 비공개 입력과 생성 산출물을 가리키며, 이 저장소만으로 실제 기업 매칭 결과를 재생성할 수는 없습니다.

## 공개 참조 실행 순서

1. `01_task_check.ipynb` - 비공개 기업 마스터 입력 계약 검증
2. `02_company_preprocessing.ipynb` - 수요기업 검색 문장 생성 (`artifacts/demand_company_embedding_input.csv`)
3. `04_technology_data_preprocessing.ipynb` - 특허·제품·솔루션 기술 문서 통합 (`artifacts/technology_embedding_input.csv`)
4. `05_technology_embedding.ipynb` - KR-SBERT 기술 임베딩·메타데이터 생성
5. `06_company_matching.ipynb` - 수요기업 문장을 임베딩하고, 기술 검색·공급기업 재정렬 결과를 생성

`06` 실행에는 위 산출물 외에 비공개 `artifacts/provider_evidence.csv`가 필요합니다. 열은 `provider_company`, `award_count`, `certificate_count`이며, 실행 결과는 `artifacts/provider_matching_result.csv`에 저장됩니다. `DEMAND_COMPANY_ID` 환경 변수로 한 수요기업을 지정합니다.

## 역사적·선택 단계

`03_homepage_description_generation.ipynb`는 홈페이지 본문으로 설명을 보강했던 2025년의 별도 검토 단계입니다. 현재 공개 참조 흐름에는 자동 결합하지 않습니다. 생성 결과는 항상 검토 전 상태이며, 현재 SDK 구현이 당시 Gemini 1.5 Flash 출력과 동일한 문장을 재현한다고 보장하지 않습니다. 결과 해석에는 `results/`의 실제 당시 산출물을 사용합니다.
