# Appendix. SciBERT 번역 실험

영어 기반 과학/특허 임베딩 모델 적용 가능성을 검토했지만, 번역 비용과 시간이 커서 최종 파이프라인에서는 제외한 실험입니다.


## 주요 단계
- 한국어 기술 설명을 영어로 번역하는 방식을 검토했습니다.
- 전체 데이터 번역 예상 시간이 프로젝트 일정에 맞지 않았습니다.
- 번역 단계를 제거할 수 있는 한국어 문장 임베딩 모델을 선택했습니다.


```python
estimated_rows = 100_000
estimated_seconds_per_row = 5.8
estimated_hours = estimated_rows * estimated_seconds_per_row / 3600
print(f'Estimated translation time: {estimated_hours:.1f} hours')
```
